```python
import numpy as np
from sklearn.preprocessing import StandardScaler, RobustScaler
import matplotlib.pyplot as plt

# Données
X = np.random.randn(100, 1)

# Ajouter outliers
X[0] = 10
X[1] = 15

# Scaling
std = StandardScaler().fit_transform(X)
rob = RobustScaler().fit_transform(X)

# Graphiques
plt.hist(X, bins=20)
plt.title("Avant scaling")
plt.show()

plt.hist(std, bins=20)
plt.title("StandardScaler (sensible aux outliers)")
plt.show()

plt.hist(rob, bins=20)
plt.title("RobustScaler (résistant)")
plt.show()
```

```python
plt.boxplot([X.flatten(), std.flatten(), rob.flatten()],
            labels=["Original", "Standard", "Robust"])
plt.title("Comparaison des effets du scaling")
plt.show()
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

X = np.array([[1,2],[3,4],[5,6]])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Données originales:\n", X)
print("Données standardisées:\n", X_scaled)
print("Moyenne:", X_scaled.mean(axis=0))
print("Variance:", X_scaled.var(axis=0))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import MaxAbsScaler

texts = [
    "I love AI",
    "AI is powerful",
    "Machine learning is amazing",
    "I love machine learning"
]

vectorizer = TfidfVectorizer()
X_sparse = vectorizer.fit_transform(texts)

print("Shape:", X_sparse.shape)
print("Nombre de valeurs non nulles:", X_sparse.nnz)

# Scaling adapté aux données sparse
scaler = MaxAbsScaler()
X_scaled = scaler.fit_transform(X_sparse)

print("Après scaling:", X_scaled.shape)
```
